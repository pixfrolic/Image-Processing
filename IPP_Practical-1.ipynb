#### 1. Write a Python program to read an image named "flower.jpg" using OpenCV and display it in a window titled "Original Image".


```python
# Load the image using OpenCV (image is loaded in BGR format).
img = cv2.imread('flower.jpg')

# Convert BGR to RGB and display the image using Matplotlib.
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')
plt.show()
```

#### 5. Write a Python program using OpenCV that resizes an image to 50% of its original width and height.

#### 6. Write a Python code snippet to crop the central 200×200 pixel area of an image and display it using Matplotlib.

```python
# Load the image using OpenCV.
img = cv2.imread('sunset.png')

# Get the height and width of the image.
h, w = img.shape[:2]

# Calculate the starting coordinates of the central crop.
x1 = (w - 200) // 2
y1 = (h - 200) // 2

# Crop the central 200 x 200 pixel area.
cropped = img[y1:y1+200, x1:x1+200]

# Display the cropped image.
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
plt.title('Cropped Image (OpenCV)')
plt.axis('off')
plt.show()
```

#### 7. Write a short program that rotates an image by 90 degrees clockwise using OpenCV.

```python
img = cv2.imread('flower.jpg')

# cv2.rotate() rotates the image 90 degrees clockwise.
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

# Display the rotated image.
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title('Rotated Image')
plt.axis('off')
plt.show()
```

#### 8. Write a Python statement to flip the image “car.jpg” left-to-right using OpenCV and display the result.

```python
img = cv2.imread('car.jpg')

# cv2.flip() flips the image horizontally.
flip_horizontal = cv2.flip(img, 1)

# Display the flipped image.
plt.imshow(cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB))
plt.title('Horizontal Flip')
plt.axis('off')
plt.show()
```

#### 9. Write a Python program that converts an image from RGB to HSV colour space, and display each channel (Hue, Saturation, Value) separately using Matplotlib.

```python
# Load the image using OpenCV.
img = cv2.imread('flower.jpg')

# Convert BGR to HSV colour space.
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Split the HSV image into three channels.
h, s, v = cv2.split(hsv)

# Display the Hue channel.
plt.subplot(1, 3, 1)
plt.imshow(h, cmap='gray')
plt.title('Hue')
plt.axis('off')

# Display the Saturation channel.
plt.subplot(1, 3, 2)
plt.imshow(s, cmap='gray')
plt.title('Saturation')
plt.axis('off')

# Display the Value channel.
plt.subplot(1, 3, 3)
plt.imshow(v, cmap='gray')
plt.title('Value')
plt.axis('off')

plt.show()
```
